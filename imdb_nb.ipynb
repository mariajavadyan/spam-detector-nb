```python
# Check for duplicates and remove if any
print(f"Duplicates in training set: {df_train.duplicated().sum()}")
df_train = df_train.drop_duplicates()
print(f"Training set size after removing duplicates: {len(df_train)}")
```

```python
import nltk

# Download the necessary NLTK data files
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

print("=== BEFORE ANY PREPROCESSING ===")
print(df_train['text'].head(5))
```

```python
import re

# Remove HTML tags (IMDB specific - contains <br /> tags)
df_train["text"] = df_train["text"].apply(lambda x: re.sub(r'<[^>]+>', ' ', x))
print("\n=== AFTER REMOVING HTML TAGS ===")
print(df_train["text"].head(5))
```

```python
# Convert all text to lowercase
df_train["text"] = df_train["text"].str.lower()
print("\n=== AFTER LOWERCASING ===")
print(df_train["text"].head(5))
```

```python
# Remove non-essential punctuation and numbers, keep useful symbols like $ and !
df_train["text"] = df_train["text"].apply(lambda x: re.sub(r"[^a-z\s$!]", "", x))
print("\n=== AFTER REMOVING PUNCTUATION & NUMBERS (except $ and !) ===")
print(df_train["text"].head(5))
```

```python
from nltk.tokenize import word_tokenize

# Split each text into individual tokens
df_train["text"] = df_train["text"].apply(word_tokenize)
print("\n=== AFTER TOKENIZATION ===")
print(df_train["text"].head(5))
```

```python
from nltk.corpus import stopwords

# Define a set of English stop words and remove them from the tokens
# IMPORTANT: Keep negation words as they carry sentiment information
stop_words = set(stopwords.words("english"))

# Preserve negation words - crucial for sentiment analysis
# These help capture patterns like "nothing happened", "never explored", "not good"
negation_words = {'not', 'no', 'never', 'nothing', 'neither', 'nobody', 'none', 'nor'}
stop_words_modified = stop_words - negation_words

df_train["text"] = df_train["text"].apply(lambda x: [word for word in x if word not in stop_words_modified])
print("\n=== AFTER REMOVING STOP WORDS (keeping negation words) ===")
print(df_train["text"].head(5))
```

```python
from nltk.stem import PorterStemmer

# Stem each token to reduce words to their base form
stemmer = PorterStemmer()
df_train["text"] = df_train["text"].apply(lambda x: [stemmer.stem(word) for word in x])
print("\n=== AFTER STEMMING ===")
print(df_train["text"].head(5))
```

```python
# Rejoin tokens into a single string for feature extraction
df_train["text"] = df_train["text"].apply(lambda x: " ".join(x))
print("\n=== AFTER JOINING TOKENS BACK INTO STRINGS ===")
print(df_train["text"].head(5))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Use TfidfVectorizer instead of CountVectorizer for better sentiment analysis
# TF-IDF weighs terms by their importance, which helps with sentiment classification
vectorizer = TfidfVectorizer(min_df=2, max_df=0.95, ngram_range=(1, 2), max_features=50000)

# Fit and transform the text column
X = vectorizer.fit_transform(df_train["text"])

# Labels (target variable) - already numeric in IMDB dataset
y = df_train["label"]

print(f"Vocabulary size: {len(vectorizer.vocabulary_)}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Build the pipeline by combining vectorization and classification
pipeline = Pipeline([
    ("vectorizer", vectorizer),
    ("classifier", MultinomialNB())
])
```

```python
# Define the parameter grid for hyperparameter tuning
# Expanded range to find optimal alpha for TF-IDF features
param_grid = {
    "classifier__alpha": [0.001, 0.005, 0.01, 0.05, 0.1, 0.15, 0.2, 0.25, 0.5, 0.75, 1.0]
}

# Perform the grid search with 5-fold cross-validation and the F1-score as metric
grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="f1"
)

# Fit the grid search on the preprocessed training data
grid_search.fit(df_train["text"], y)

# Extract the best model identified by the grid search
best_model = grid_search.best_estimator_
print("Best model parameters:", grid_search.best_params_)
print(f"Best cross-validation F1 score: {grid_search.best_score_:.4f}")
```

```python
# Preprocess function that mirrors the training-time preprocessing
def preprocess_text(text):
    # Remove HTML tags
    text = re.sub(r'<[^>]+>', ' ', text)
    # Lowercase
    text = text.lower()
    # Remove punctuation and numbers except $ and !
    text = re.sub(r"[^a-z\s$!]", "", text)
    # Tokenize
    tokens = word_tokenize(text)
    # Remove stopwords (but keep negation words)
    tokens = [word for word in tokens if word not in stop_words_modified]
    # Stem
    tokens = [stemmer.stem(word) for word in tokens]
    return " ".join(tokens)
```

```python
# Preprocess the test data
df_test["text_processed"] = df_test["text"].apply(preprocess_text)
print("Test data preprocessed.")
print(df_test["text_processed"].head(5))
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Predict on preprocessed test data
y_pred = best_model.predict(df_test["text_processed"])

print(f"Test Accuracy: {accuracy_score(df_test['label'], y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(df_test['label'], y_pred, target_names=['Negative', 'Positive']))
```

```python
# Example movie reviews for evaluation
new_reviews = [
    "This movie was absolutely fantastic! The acting was superb and the plot kept me engaged throughout.",
    "Terrible film. Waste of time and money. The worst movie I've ever seen.",
    "An average movie, nothing special but not bad either. It was okay.",
    "A masterpiece of cinema! The director did an amazing job with this beautiful story.",
    "Boring and predictable. I fell asleep halfway through. Do not recommend.",
]
```

```python
# Preprocess and predict on new reviews
processed_reviews = [preprocess_text(review) for review in new_reviews]

# Predict with the trained classifier
predictions = best_model.predict(processed_reviews)
prediction_probabilities = best_model.predict_proba(processed_reviews)

# Display predictions and probabilities for each evaluated review
for i, review in enumerate(new_reviews):
    prediction = "Positive" if predictions[i] == 1 else "Negative"
    positive_probability = prediction_probabilities[i][1]
    negative_probability = prediction_probabilities[i][0]
    
    print(f"Review: {review}")
    print(f"Prediction: {prediction}")
    print(f"Positive Probability: {positive_probability:.2f}")
    print(f"Negative Probability: {negative_probability:.2f}")
    print("-" * 50)
```
